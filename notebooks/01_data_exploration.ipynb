# Data Exploration

Load the demo datasets and inspect the structure.

```python
import pandas as pd
from pathlib import Path

base = Path('..') / 'data' / 'raw'
trends = pd.read_csv(base / 'google_trends' / 'google_trends_demo.csv', parse_dates=['date'])
search = pd.read_csv(base / 'statcounter' / 'search_engine_share_demo.csv', parse_dates=['date'])
browser = pd.read_csv(base / 'statcounter' / 'browser_share_demo.csv', parse_dates=['date'])
ai = pd.read_csv(base / 'similarweb' / 'ai_tool_traffic_demo.csv', parse_dates=['date'])
events = pd.read_csv(base / 'events' / 'events.csv', parse_dates=['event_date'])

print(trends.head())
print(search.head())
print(browser.head())
print(ai.head())
print(events.head())
```
