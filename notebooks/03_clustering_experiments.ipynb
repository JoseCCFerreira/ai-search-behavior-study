# Clustering Experiments

Create clustering prototypes for countries and platforms.

```python
import pandas as pd
from pathlib import Path
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans

base = Path('..') / 'data' / 'raw'
ai = pd.read_csv(base / 'similarweb' / 'ai_tool_traffic_demo.csv', parse_dates=['date'])
sample = ai.groupby(['region','ai_tool'])['traffic_share_percent'].mean().reset_index()
features = sample.pivot(index='region', columns='ai_tool', values='traffic_share_percent').fillna(0)
model = KMeans(n_clusters=3, random_state=42).fit(features)
pca = PCA(n_components=2).fit_transform(features)
print(features.head())
print(model.labels_)
```
