# Forecasting Experiments

Try baseline and moving average forecasting approaches.

```python
import pandas as pd
from pathlib import Path

base = Path('..') / 'data' / 'raw'
ai = pd.read_csv(base / 'similarweb' / 'ai_tool_traffic_demo.csv', parse_dates=['date'])
sample = ai[ai['ai_tool'] == 'ChatGPT'].groupby('date')['traffic_share_percent'].mean().reset_index()
sample['forecast'] = sample['traffic_share_percent'].rolling(window=3, min_periods=1).mean()
print(sample.tail())
```
