# Feature Engineering

Create time-based features and rolling metrics.

```python
import pandas as pd
from pathlib import Path

base = Path('..') / 'data' / 'raw'
trends = pd.read_csv(base / 'google_trends' / 'google_trends_demo.csv', parse_dates=['date'])
trends['month'] = trends['date'].dt.to_period('M')
trends['interest_mom'] = trends.groupby(['region','keyword'])['interest'].pct_change()
trends['interest_yoy'] = trends.groupby(['region','keyword'])['interest'].pct_change(12)
print(trends[['date','region','keyword','interest','interest_mom','interest_yoy']].head())
```
